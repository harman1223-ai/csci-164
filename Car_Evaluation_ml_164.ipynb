### Dataset Description and Structure

The dataset used is the "Car Evaluation" Dataset from the UCI Machine Learning Repository. It contains 1,728 instances** and 7 attributes,  6 input features and 1 target variable.

### Features:
All features are categorical:
- buying – buying price: `vhigh`, `high`, `med`, `low`
- maint – maintenance cost: `vhigh`, `high`, `med`, `low`
- doors – number of doors: `2`, `3`, `4`, `5more`
- persons – capacity in terms of poeple to carry: `2`, `4`, `more`
- lug_boot – size of luggage boot: `small`, `med`, `big`
- safety – safety rating: `low`, `med`, `high`

### Target Variable:
- class – car acceptability: one of `unacc`, `acc`, `good`, or `vgood`

### Data Types:
All variables are categorical and were label-encoded for model compatibility.

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

```python
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/car/car.data"
column_names = ['buying', 'maint', 'doors', 'persons', 'lug_boot', 'safety', 'class']

df = pd.read_csv(url, names=column_names)
df.head()
```

```python
print("Shape of dataset:", df.shape)
print("\nColumn types:")
print(df.dtypes)
```

```python
df_encoded = df.copy()
le = LabelEncoder()

for col in df_encoded.columns:
    df_encoded[col] = le.fit_transform(df_encoded[col])

df_encoded.head()
```

```python
from sklearn.model_selection import train_test_split

# Features (X) and target (y)
X = df_encoded.drop('class', axis=1)
y = df_encoded['class']

# Split into training and testing
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)
```

```python
from sklearn.linear_model import LogisticRegression

# Train the model
clf = LogisticRegression(max_iter=200)
clf.fit(X_train, y_train)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix

# Predictions
y_pred = clf.predict(X_test)

# Accuracy
print("Accuracy:", accuracy_score(y_test, y_pred))

# Confusion matrix
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
```

```python
from sklearn.neighbors import KNeighborsClassifier

# Initialize and train k-NN (k=5)
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)

# Predict
y_pred_knn = knn.predict(X_test)
```

```python
# Accuracy
print("k-NN Accuracy:", accuracy_score(y_test, y_pred_knn))

# Confusion matrix
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred_knn))
```

# Classification Results – k-Nearest Neighbors (k=5)

I applied a k-NN classifier with k=5. Like with logistic regression, we trained on 80% of the dataset and tested on the remaining 20%

Evaluation
- The accuracy score indicates overall correctness
- The confusion matrix shows per-class performance
- 
This model will be compared to the logistic regression model and others in the evaluation section


```python
# Manual mapping of class labels to ordinal values
label_mapping = {'unacc': 0, 'acc': 1, 'good': 2, 'vgood': 3}
df_regression = df.copy()
df_regression['class'] = df_regression['class'].map(label_mapping)

# Encode features for regression
le = LabelEncoder()
for col in df_regression.columns[:-1]:
    df_regression[col] = le.fit_transform(df_regression[col])
```

```python
from sklearn.linear_model import LinearRegression

# Features and target
X_reg = df_regression.drop('class', axis=1)
y_reg = df_regression['class']

# Split
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(X_reg, y_reg, test_size=0.2, random_state=1)

# Train linear regression
reg = LinearRegression()
reg.fit(X_train_reg, y_train_reg)

# Predict
y_pred_reg = reg.predict(X_test_reg)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Evaluation metrics
mae = mean_absolute_error(y_test_reg, y_pred_reg)
mse = mean_squared_error(y_test_reg, y_pred_reg)
rmse = np.sqrt(mse)
r2 = r2_score(y_test_reg, y_pred_reg)

print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R² Score:", r2)
```

### Model Choices

I used three models from scikit-learn:

- Logistic Regression: Good starting point for multiclass classification, easy to use and fast
- k-Nearest Neighbors(k-NN): Works well with label-encoded data and doesn’t assume anything about the data
- Linear Regression: Used after mapping the class labels to numbers so we can try predicting a score

All models were trained on 80% of the data and tested on 20% and classification models were checked with accuracy and F1-score, and regression was checked with MAE, MSE, and R².



```python
from sklearn.cluster import KMeans

# Use the encoded features only (drop the label)
X_cluster = df_encoded.drop('class', axis=1)
```

```python
# Apply KMeans with 4 clusters (since there are 4 class labels)
kmeans = KMeans(n_clusters=4, random_state=1)
kmeans.fit(X_cluster)

# Predicted clusters
clusters = kmeans.labels_
```

```python
# Add clusters and original class back to a new DataFrame
df_clustered = df_encoded.copy()
df_clustered['cluster'] = clusters

# Cross-tabulation to see how clusters align with actual class labels
pd.crosstab(df_clustered['cluster'], df_encoded['class'])
```

```python
from sklearn.model_selection import GridSearchCV

# Parameter grid for k-NN
param_grid_knn = {
    'n_neighbors': [3, 5, 7, 9],
    'weights': ['uniform', 'distance']
}
```

```python
# Re-initialize model
knn = KNeighborsClassifier()

# Set up grid search
grid_knn = GridSearchCV(knn, param_grid_knn, cv=5)
grid_knn.fit(X_train, y_train)

# Best parameters and score
print("Best Parameters:", grid_knn.best_params_)
print("Best Cross-Validated Score:", grid_knn.best_score_)
```

```python
# Use the best model to predict test set
y_pred_best_knn = grid_knn.predict(X_test)

# Evaluate
print("Test Accuracy (Best k-NN):", accuracy_score(y_test, y_pred_best_knn))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred_best_knn))
```

### Hyperparameter Tuning

We used "GridSearchCV" to tune the k-Nearest Neighbors (k-NN) classifier.  
The parameters we tested were:

- n_neighbors: 3, 5, 7, 9  
- weights: 'uniform', 'distance'

The best combination found was n_neighbors = 7 and weights = 'uniform'  
This model had a cross-validated score of about 90.89% and performed better than the default k-NN and Logistic Regression models.
